```python
def setup_unity_catalog(catalog: str, schemas: list[str]):
    """
    Realiza o bootstrap do Unity Catalog e Schemas com metadados.
    """

    # Creating catalog
    spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog} "
              f"COMMENT 'Catalogo de dominio para o case NYC Taxi - iFood'")

    for schema in schemas:
        # Creating schemas
        comment = f"Camada {schema.upper()} para o dominio {catalog}"
        
        spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema} "
                  f"COMMENT '{comment}'")
        
    print(f"Setup concluído: Catálogo '{catalog}' e schemas {schemas} prontos.")
```

```python
CATALOG = "logistics"
SCHEMAS = ["bronze", "silver", "gold", "quarantine"]

setup_unity_catalog(catalog=CATALOG, schemas=SCHEMAS)
```
